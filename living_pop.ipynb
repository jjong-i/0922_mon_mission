```python
import pandas as pd
import os

# 1. 통합할 연도별 파일 명단 (월별 파일 제외하고 딱 8개만 지정)
years = range(2017, 2025)
target_files = [f"LOCAL_PEOPLE_GU_{year}.csv" for year in years]

print(f"🚀 [최종] 8개 연도 파일 병합 시작! (대상: {target_files[0]} ~ {target_files[-1]})\n")

all_data = []
standard_columns = None # 2017년(기준) 컬럼명 저장용

for file in target_files:
    if not os.path.exists(file):
        print(f"⚠️ 파일 없음: {file} (중단)")
        break

    print(f"📄 처리 중: {file} ...", end=" ")

    try:
        # [읽기] 인코딩 자동 감지 (utf-8-sig 우선 시도 -> 실패 시 cp949)
        try:
            df = pd.read_csv(file, encoding='utf-8-sig')
        except:
            df = pd.read_csv(file, encoding='cp949')

        # [A] 2017년: 기준 컬럼명 저장
        if "2017" in file:
            standard_columns = df.columns
            print(f"-> [기준 컬럼 확보]", end=" ")

        # [B] 2024년: 수술 필요 (컬럼 삭제 + 이름 변경)
        elif "2024" in file:
            # 1. 적재일시 컬럼 삭제
            if 'ldadng_dt' in df.columns:
                df = df.drop(columns=['ldadng_dt'])
                print("(적재일시 삭제됨)", end=" ")

            # 2. 영어 컬럼 -> 한글 기준 컬럼으로 덮어쓰기
            if standard_columns is not None and len(df.columns) == len(standard_columns):
                df.columns = standard_columns
                print("-> [한글 변환 완료]", end=" ")
            else:
                print(f"-> ⚠️ 경고: 컬럼 수 안 맞음({len(df.columns)})", end=" ")

        # [C] 2018~2023년: 그냥 통과 (이미 정상이므로)
        else:
            # 혹시 모를 안전장치: 컬럼명 강제 통일
            if standard_columns is not None:
                df.columns = standard_columns
            print("-> 정상 로드", end=" ")

        # 리스트에 담기
        all_data.append(df)
        print(f"({len(df):,}행)")

    except Exception as e:
        print(f"\n❌ 에러 발생: {e}")

# --- 최종 병합 및 저장 ---
if len(all_data) == 8:
    print("\n📦 데이터 8년 치 합치는 중... (잠시만요)")
    final_df = pd.concat(all_data, axis=0, ignore_index=True)

    save_name = "SEOUL_LIVING_POPULATION_2017_2024.csv"
    final_df.to_csv(save_name, index=False, encoding='utf-8-sig')

    print("\n" + "="*40)
    print(f"🎉 축하합니다! 통합 작업이 완벽하게 끝났습니다.")
    print(f"📂 생성된 파일: {save_name}")
    print(f"📊 총 데이터 행: {len(final_df):,} 개")
    print("="*40)

    # 마지막 검증: 2024년 12월 31일 데이터가 보이는지 확인
    print("\n[최종 데이터 확인 - 끝부분]")
    print(final_df.tail())
else:
    print(f"\n⚠️ 파일이 8개가 다 모이지 않았습니다. (현재 {len(all_data)}개)")
```
